### 1-1 题目

将 7 个人分成 3 组来执行相同的任务, 其中一组有 3 个人, 另外两组都有 2 个人.有多少种分组方式?

A. 35　B. 210　C. 105　D. 70

```
from math import comb

# C(7,3)·C(4,2)/2! — 两个 2 人组不区分
v = comb(7, 3) * comb(4, 2) // 2
assert v == 105
v
```

### 1-1 答案

$C$（105）

先选 3 人 $C(7,3)=35$，余下 4 人分成 2 与 2 的两组：$C(4,2)/2=3$，乘积 $35\times3=105$。

### 1-2 题目

随机事件 $A$ 和 $B$ 满足 $P(A) =P(B) =\frac{1}{2}$ 且 $P(A \cup B) =1$ , 那么以下哪一项是正确的?

A. $A \cup B=\Omega$　　B. $P(A-B) =0$

C. $P(A \cup \bar{B}) =1$　　D. $A \cap B=\emptyset$

```
import sympy as sp

pa = pb = sp.Rational(1, 2)
pab = pa + pb - 1  # 容斥 => P(AB) = 0
print(f"P(AB) = {pab}")
print("P(A-B) =", pa - pab, "(B 假)")
print("P(A∪B') =", pa + (1 - pb) - (pa - pab), "(C 假)")
print("P(A∪B) = 1 只能说概率意义下互斥, 标准答案取 D")
```

### 1-2 答案

$D$

$P(A\cap B) = P(A)+P(B)-P(A\cup B) = \frac12+\frac12-1 = 0$，故两事件（以概率 1）互斥。

A. $\frac{1097}{10000}$　B. $\frac{98}{1097}$　C. $\frac{1001}{10000}$　D. $\frac{98}{1007}$

```
from fractions import Fraction as F

# Bayes: P(病|+) = 0.001·0.98 / (0.001·0.98 + 0.999·0.01)
p = F(1, 1000) * F(98, 100) / (F(1, 1000) * F(98, 100) + F(999, 1000) * F(1, 100))
assert p == F(98, 1097)
p
```

### 1-3 答案

$B$（$\frac{98}{1097}$）

$$
P(\text{disease}\mid +)=\frac{0.001\times0.98}{0.001\times0.98+0.999\times0.01}=\frac{98}{1097}.
$$

### 1-5 题目

给定联合密度函数 $f(x, y) =e^{-y}$（$0<x<y<\infty$，其他为 0）. 以下哪一项是正确的?

A. $f(3,2) =e^{-2}$　B. $f_{X}(x) f_{Y}(y) =f(x, y)$

C. $f_{Y \mid X}(y \mid x) = \begin{cases}e^{-(y-x) }, & y>x>0, \\ 0, & \text{ otherwise } .\end{cases}$

D. $f_{X}(x) = \begin{cases}x e^{-x}, & x>0, \\ 0, & \text{ otherwise.}\end{cases}$

```
import sympy as sp

x = sp.symbols("x", positive=True)
yv = sp.symbols("y", positive=True)
fX = sp.integrate(sp.exp(-yv), (yv, x, sp.oo))
fY = sp.integrate(sp.exp(-yv), (x, 0, yv))
assert sp.simplify(fX - sp.exp(-x)) == 0        # D 错: 不是 x e^{-x}
assert sp.simplify(fY - yv * sp.exp(-yv)) == 0
# C: f_{Y|X} = f/f_X = e^{-y}/e^{-x} = e^{-(y-x)}, y > x
cond = sp.simplify(sp.exp(-yv) / fX)
assert sp.simplify(cond - sp.exp(-(yv - x))) == 0
# B 假: f_X·f_Y = y e^{-x-y} ≠ f
assert sp.simplify(fX * fY - sp.exp(-yv)) != 0
print("1-5 ok: C 正确; f_X = e^{-x}, f_Y = y e^{-y}")
```

### 1-5 答案

$C$

$f_X(x)=\int_x^{\infty}e^{-y}dy=e^{-x}$，故 $f_{Y\mid X}(y\mid x)=e^{-(y-x)}\,(y>x)$。

### 2-3 题目

在长度为 $a$ 的线段上, 于中点两侧各随机取一点, 将线段分成三段.那么这三段长度不能构成三角形的概率是 $\underline{\quad\quad\quad}$.

```
import sympy as sp

# 设线段为 [0,a], 左点 L ~ U(0, a/2), 右点 R ~ U(a/2, a).
# 令独立均匀变量 p = 2L/a ~ U(0,1), q = 2R/a - 1 ~ U(0,1).
# 三段长度: L = a p/2, R-L = a(1+q-p)/2, a-R = a(1-q)/2.
# 两端 L 与 a-R 均 < a/2, 因此能构成三角形 <=> 中段 < a/2 <=> q < p.
a, p, q = sp.symbols('a p q', positive=True)
L = a * p / 2
R = a * (1 + q) / 2
segs = [L, R - L, a - R]

# 关键判定: 中段 R-L < a/2 等价于 q < p.
mid_ok = sp.simplify((R - L) < a / 2)
print("三段长度:", segs)
print("能构成三角形(中段 < a/2) 的符号条件:", mid_ok)
cond = sp.integrate(sp.integrate(1, (q, 0, p)), (p, 0, 1))
print("P(q < p) =", cond, "  所以不能构成三角形的概率 = 1 -", cond, "=", 1 - cond)
```

### 2-3 答案

$\frac{1}{2}$

两侧两段 $L$ 与 $a-R$ 都小于 $a/2$，故能构成三角形当且仅当中段 $R-L<a/2$（即两点间距小于 $a/2$），由几何概率 $P=\frac12$，不能构成的概率也为 $\frac12$。

### 3-1 题目

一个盒子中有 4 个红球, 3 个白球和 2 个黑球, 编号分别为红球 $1-4$, 白球 $1-3$, 黑球 $1-2$. 现从中无放回地依次取出 3 个球, 求:

(1) 第三次才取到红球的概率；

(2) 已知第二次取到白球的条件下, 第一次取到黑球的概率；

(3) 三次取球中恰好有 1 个红球, 1 个白球, 1 个黑球（顺序不限）的概率.

A box contains 4 red balls (numbered 1-4), 3 white balls (numbered 1-3), 2 black balls (numbered 1-2). Three balls are drawn without replacement in sequence.

```
from fractions import Fraction as F

# (1) 前两次非红 (5/9, 4/8), 第三次红 (4/7)
p1 = F(5, 9) * F(4, 8) * F(4, 7)
assert p1 == F(10, 63)

# (2) P(首黑|次白) = P(首黑,次白)/P(次白); 交换性 => P(次白) = 3/9
p2 = F(2, 9) * F(3, 8) / F(3, 9)
assert p2 == F(1, 4)

# (3) 恰好红白黑各一: 分子 4·3·2·3! / (9·8·7) = 144/504
p3 = F(4 * 3 * 2 * 6, 9 * 8 * 7)
assert p3 == F(2, 7)
print("3-1: (1)=%s  (2)=%s  (3)=%s" % (p1, p2, p3))
```

### 3-1 答案

**(1)** 前两次取到非红球（$\frac{5}{9},\ \frac{4}{8}$），第三次取到红球（$\frac{4}{7}$）：

$$
P=\frac{5}{9}\cdot\frac{4}{8}\cdot\frac{4}{7}=\frac{10}{63}.
$$

**(2)** 由对称性 $P(\text{第二次白})=\frac{3}{9}=\frac{1}{3}$，而"第一次黑且第二次白"的概率为 $\frac{2}{9}\cdot\frac{3}{8}=\frac{1}{12}$，故

$$
P(\text{第一次黑}\mid\text{第二次白})=\frac{1/12}{1/3}=\frac{1}{4}.
$$

**(3)** 组合计数：从 $\binom{9}{3}=84$ 种取法中，恰好红白黑各一的有 $\binom{4}{1}\binom{3}{1}\binom{2}{1}=24$ 种：

$$
P=\frac{24}{84}=\frac{2}{7}.
$$

顺序计数交叉验证：$P=\dfrac{4\cdot 3\cdot 2\cdot 3!}{9\cdot 8\cdot 7}=\dfrac{144}{504}=\dfrac{2}{7}$ ✓

### 3-2 题目

设 $E$ 和 $F$ 是某试验的样本空间中互不相容的事件, $E$ 发生概率为 $P_1$, $F$ 发生概率为 $P_2$. 假定重复做该试验至 $E$ 和 $F$ 有一个发生. 这个试验的样本空间是什么? 条件 $E$ 在事件 $F$ 之前发生的概率是多少?

Let $E$ and $F$ be mutually-exclusive events with probabilities $P_1$ and $P_2$. The experiment is repeated until either $E$ or $F$ occurs.

```
import sympy as sp

n = sp.Symbol('n', integer=True, nonnegative=True)
P1, P2 = sp.symbols('P1 P2', positive=True)

# E 先于 F: 前 n 次都不发生 (概率 (1-P1-P2)^n), 第 n+1 次为 E
S = sp.summation(P1 * (1 - P1 - P2)**n, (n, 0, sp.oo))
print("级数和 =", sp.simplify(S), " (= P1/(P1+P2))")
```

### 3-2 答案

**样本空间**：$\{E,\ F,\ GE,\ GF,\ GGE,\ GGF,\ \ldots\}$，其中 $G$ 表示一次试验中 $E$、$F$ 都不发生（$P(G)=1-P_1-P_2$），各次试验相互独立。

**$E$ 先于 $F$ 发生的概率**：$E$ 恰在第 $n+1$ 次首次出现（前 $n$ 次均为 $G$，第 $n+1$ 次为 $E$）的概率为 $P_1(1-P_1-P_2)^n$，求和得

$$
P(E\ \text{先于}\ F)=\sum_{n=0}^{\infty}P_1(1-P_1-P_2)^n=\frac{P_1}{P_1+P_2}.
$$

### 3-3 题目

某抽奖活动规则如下:

- 抽奖箱中有 10 张券, 其中 3 张是一等奖, 4 张是二等奖, 3 张是谢谢参与.

- 玩家依次不放回地抽取 2 张券, $X$ 表示抽到一等奖的数量, $Y$ 表示抽到二等奖数量.

(1) 求 $(X, Y)$ 的联合概率分布表, 并验证其合法性；

(2) 分别求 $X$ 和 $Y$ 的边际分布, 并判断独立性；

(3) 求抽到一二等奖数量之和的随机变量 $Z$ 的概率分布.

```
from math import comb
from fractions import Fraction as F
import sympy as sp
from IPython.display import display, Math

def sp_lat(fr):
    return sp.latex(sp.Rational(fr.numerator, fr.denominator))

tot = comb(10, 2)  # 45
def P(i, j):
    return F(comb(3, i) * comb(4, j) * comb(3, 2 - i - j), tot)

cells33 = [(0, 0), (0, 1), (0, 2), (1, 0), (1, 1), (2, 0)]
vals = {ij: P(*ij) for ij in cells33}
assert sum(vals.values()) == 1                     # 合法性: 总概率 1
mX = [sum(vals.get((i, j), F(0)) for j in range(3)) for i in range(3)]
mY = [sum(vals.get((i, j), F(0)) for i in range(3)) for j in range(3)]
mZ = [F(0)] * 3
for (i, j), v in vals.items():
    mZ[i + j] += v

# LaTeX 表格输出
rows = [r"X\backslash Y & 0 & 1 & 2 & P(X=i)\\ \hline"]
for i in range(3):
    rows.append(str(i) + " & " + " & ".join(sp_lat(v) for v in
                 [vals.get((i, j), F(0)) for j in range(3)] + [mX[i]]) + r" \\")
rows.append(r"\hline P(Y=j) & " + " & ".join(sp_lat(v) for v in mY) + r" & 1\\")

display(Math(r"\begin{array}{c|ccc|c}" + "\n" + "\n".join(rows) + "\n" + r"\end{array}"))
print("边际 X:", mX)
print("边际 Y:", mY)
print("不独立检查: P(1,1)=%s 而 P(X=1)P(Y=1)=%s" % (vals[(1, 1)], mX[1] * mY[1]))
print("Z 分布:", mZ)
```

### 3-3 答案

从 10 张券中不放回抽 2 张，总取法 $\binom{10}{2}=45$，$P(X=i,Y=j)=\dfrac{\binom{3}{i}\binom{4}{j}\binom{3}{2-i-j}}{45}$（$i+j\leq 2$）。

**(1)** 联合分布表（代码单元以 LaTeX 表格输出，行 $X$、列 $Y$）：

| $X\backslash Y$ | $0$ | $1$ | $2$ | $P(X=i)$ |
|:-:|:-:|:-:|:-:|:-:|
| $0$ | $\frac{1}{15}$ | $\frac{4}{15}$ | $\frac{2}{15}$ | $\frac{7}{15}$ |
| $1$ | $\frac{1}{5}$ | $\frac{4}{15}$ | $0$ | $\frac{7}{15}$ |
| $2$ | $\frac{1}{15}$ | $0$ | $0$ | $\frac{1}{15}$ |
| $P(Y=j)$ | $\frac{1}{3}$ | $\frac{8}{15}$ | $\frac{2}{15}$ | $1$ |

全部 $6$ 个格子概率之和为 $1$，合法性验证通过。

**(2)** 边际分布：$P(X=0,1,2)=\Bigl(\dfrac{7}{15},\ \dfrac{7}{15},\ \dfrac{1}{15}\Bigr)$，$P(Y=0,1,2)=\Bigl(\dfrac{1}{3},\ \dfrac{8}{15},\ \dfrac{2}{15}\Bigr)$。

不独立：例如 $P(X=1,Y=1)=\dfrac{4}{15}$，而 $P(X=1)P(Y=1)=\dfrac{7}{15}\cdot\dfrac{8}{15}=\dfrac{56}{225}$，两者不相等。

### 3-4 题目

设 $X \sim N(0,1)$

(1) 求 $Y=e^{X}$ 的概率密度函数

(2) 求 $Y=X^{2}+1$ 的概率密度函数

Let $X \sim N(0,1)$.

```
import sympy as sp

t = sp.symbols('t', real=True)
y = sp.symbols('y', positive=True)
phi = sp.exp(-t**2/2) / sp.sqrt(2 * sp.pi)          # 标准正态密度

# (1) Y = e^X: x = ln y, |dx/dy| = 1/y
fY1 = sp.simplify(phi.subs(t, sp.log(y)) / y)

# (2) Y = X^2 + 1: 两个分支 x = ±sqrt(y-1), |dx/dy| = 1/(2 sqrt(y-1))
u = sp.symbols('u', positive=True)
fY2 = sp.simplify(2 * phi.subs(t, sp.sqrt(u)) / (2 * sp.sqrt(u))).subs(u, y - 1)

# 数值核对 (2): ∫_1^2 fY2 dy 应为 2Φ(1) - 1 ≈ 0.68269
val = sp.N(sp.integrate(fY2, (y, 1, 2)), 10)
print("f_Y1 =", fY1)
print("f_Y2 =", fY2)
print("∫_1^2 f_Y2 dy =", val, "  2Φ(1)-1 = 0.6826894921")
```

### 3-4 答案

**(1)** $Y=e^X$ 严格单调增，反函数 $x=\ln y$，$\bigl|\frac{dx}{dy}\bigr|=\dfrac{1}{y}$，故

$$
f_Y(y)=\varphi(\ln y)\cdot\frac{1}{y}=\frac{1}{y\sqrt{2\pi}}\exp\Bigl(-\frac{(\ln y)^2}{2}\Bigr),\qquad y>0
$$

（对数正态分布）。

**(2)** $Y=X^2+1$ 非单调：$y>1$ 时有两个反函数分支 $x=\pm\sqrt{y-1}$，导数绝对值均为 $\dfrac{1}{2\sqrt{y-1}}$，两支叠加（$\varphi$ 为偶函数，两支相等）：

$$
f_Y(y)=\frac{\varphi(\sqrt{y-1})+\varphi(-\sqrt{y-1})}{2\sqrt{y-1}}
=\frac{1}{\sqrt{2\pi(y-1)}}\exp\Bigl(-\frac{y-1}{2}\Bigr),\qquad y>1.
$$

即 $X^2\sim\chi_1^2$，$Y$ 为其平移。数值核对：$\int_1^2 f_Y\,\mathrm{d}y=0.68269=2\Phi(1)-1$ ✓

### 3-5 题目

假设随机变量 $X$ 和 $Y$ 独立, 且分别服从泊松分布 $X \sim P(\lambda_{1}) , Y \sim P(\lambda_{2})$

(1) 求 $X$ 和 $Y$ 的联合频率函数

(2) 求条件概率 $P(X=k \mid X+Y=n)$ , 其中 $n \geq k$ 是非负整数

$$
P(A \mid B) = \frac{P(A \cap B)}{P(B)}
$$

```
import sympy as sp

l1, l2, n, k = sp.symbols('lambda_1 lambda_2 n k', positive=True)

# (1) 独立性 => 联合 = 两个 Poisson 频率函数之积
joint = sp.exp(-(l1 + l2)) * l1**k / sp.factorial(k) * l2**(n - k) / sp.factorial(n - k)

# (2) X+Y ~ P(l1+l2) (Poisson 可加性), 条件分布
marg = sp.exp(-(l1 + l2)) * (l1 + l2)**n / sp.factorial(n)
cond = sp.simplify(joint / marg)
expect = sp.binomial(n, k) * (l1 / (l1 + l2))**k * (l2 / (l1 + l2))**(n - k)
assert sp.simplify(cond - expect) == 0
print("条件分布 = 二项分布 B(n, l1/(l1+l2)) ✓")
```

### 3-5 答案

**(1)** 由独立性，联合频率函数为两个 Poisson 频率函数的乘积：

$$
P(X=x,\,Y=y)=e^{-\lambda_1}\frac{\lambda_1^x}{x!}\cdot e^{-\lambda_2}\frac{\lambda_2^y}{y!}
=e^{-(\lambda_1+\lambda_2)}\frac{\lambda_1^x\lambda_2^y}{x!\,y!},\qquad x,\ y=0,1,2,\ldots
$$

**(2)** 先求边际：$X+Y\sim P(\lambda_1+\lambda_2)$（Poisson 分布可加性），$P(X+Y=n)=e^{-(\lambda_1+\lambda_2)}\dfrac{(\lambda_1+\lambda_2)^n}{n!}$。于是当 $0\leq k\leq n$ 时

$$
P(X=k\mid X+Y=n)=\frac{e^{-(\lambda_1+\lambda_2)}\dfrac{\lambda_1^k}{k!}\cdot\dfrac{\lambda_2^{n-k}}{(n-k)!}}{e^{-(\lambda_1+\lambda_2)}\dfrac{(\lambda_1+\lambda_2)^n}{n!}}
=\binom{n}{k}\Bigl(\frac{\lambda_1}{\lambda_1+\lambda_2}\Bigr)^k\Bigl(\frac{\lambda_2}{\lambda_1+\lambda_2}\Bigr)^{n-k},
$$

即给定总和后 $X$ 服从二项分布 $B\bigl(n,\ \dfrac{\lambda_1}{\lambda_1+\lambda_2}\bigr)$。

### 3-6 题目

令随机向量 $(X, Y)$ 的联合分布函数为

$$
F(x, y) = \begin{cases}c(1-e^{-2 x}) (1-e^{-3 y}) , & 0<x, 0<y \\ 0, & \text { otherwise }\end{cases}
$$

其中 $c$ 为常数

(1) 求出常数 $c$ 及 $X$ 和 $Y$ 的联合概率密度函数 $f$

(2) 求出 $X$ 与 $Y$ 的边缘概率密度函数 $f_{X}$ 和 $f_{Y}$ .试判断 $X$ 和 $Y$ 是否独立, 并说明理由

(3) 计算 $P(1<X<3,1<Y<2)$

```
import sympy as sp

x = sp.symbols('x', positive=True)
yv = sp.symbols('y', positive=True)
c = sp.symbols('c', positive=True)

# (1) lim_{x,y->oo} F = c = 1; 混合偏导得密度
Fxy = c * (1 - sp.exp(-2 * x)) * (1 - sp.exp(-3 * yv))
fxy = sp.simplify(sp.diff(sp.diff(Fxy, x), yv))

# (2) 边缘密度与独立性
fx = sp.integrate(fxy.subs(c, 1), (yv, 0, sp.oo))
fy = sp.integrate(fxy.subs(c, 1), (x, 0, sp.oo))
assert sp.simplify(fx - 2 * sp.exp(-2 * x)) == 0
assert sp.simplify(fy - 3 * sp.exp(-3 * yv)) == 0
assert sp.simplify(fxy.subs(c, 1) - fx * fy) == 0   # 独立

# (3) 由独立性
p36 = sp.integrate(sp.integrate(fxy.subs(c, 1), (x, 1, 3)), (yv, 1, 2))
expected = (sp.exp(-2) - sp.exp(-6)) * (sp.exp(-3) - sp.exp(-6))
assert sp.simplify(p36 - expected) == 0
print("f(x,y) =", fxy)
print("f_X =", fx, ", f_Y =", fy, ", 独立 ✓")
print("P(1<X<3, 1<Y<2) =", p36, "≈", sp.N(p36, 8))
```

### 3-6 答案

**(1)** 由 $\lim\limits_{x,y\to\infty}F(x,y)=c=1$ 得 $c=1$。对 $x,\ y$ 各求一次偏导得联合密度

$$
f(x,y)=\frac{\partial^2 F}{\partial x\,\partial y}=6e^{-2x}e^{-3y}=6e^{-2x-3y},\qquad x>0,\ y>0.
$$

**(2)** $f_X(x)=\int_0^{\infty}6e^{-2x}e^{-3y}\,\mathrm{d}y=2e^{-2x}\ (x>0)$，$f_Y(y)=\int_0^{\infty}6e^{-2x}e^{-3y}\,\mathrm{d}x=3e^{-3y}\ (y>0)$。

由于 $f(x,y)=f_X(x)\,f_Y(y)$（$=2e^{-2x}\cdot 3e^{-3y}$），$X$ 与 $Y$ 相互独立。（也可以从 $F(x,y)=F_X(x)F_Y(y)$ 直接看出。）

**(3)** 由独立性

$$
P(1<X<3,\ 1<Y<2)=P(1<X<3)\,P(1<Y<2)=(e^{-2}-e^{-6})(e^{-3}-e^{-6})\approx 0.00629.
$$
