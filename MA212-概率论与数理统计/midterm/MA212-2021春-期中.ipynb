## 三 解答题

### 3-1 题目

A student needs to select 1 choice from 4 optional choices for answering a question.

The student will randomly select a choice if having not got the answer, and of course will select it if having obtained the answer.

Now the student's selected choice is the answer, what is the probability that the student has obtained the answer before selecting a choice based on the following scenarios?

+ (a) The probability of having obtained the answer before selecting a choice is 0.5.
+ (b) The probability of having obtained the answer before selecting a choice is 0.2

```python
from fractions import Fraction as F

# Bayes: P(B|A) = P(B)·1 / (P(B)·1 + (1-P(B))·1/4)
b1 = F(1, 2) * 1 / (F(1, 2) * 1 + F(1, 2) * F(1, 4))
b2 = F(1, 5) * 1 / (F(1, 5) * 1 + F(4, 5) * F(1, 4))
assert b1 == F(4, 5) and b2 == F(1, 2)
(float(b1), float(b2))
```

### 3-1 答案

设事件 $A$ 为“答对”，$B$ 为“事先知道正确答案”。则 $P(A\mid B)=1,\ P(A\mid \bar B)=\tfrac14$。

(a) 先验 $P(B)=0.5$，有
$$P(B\mid A)=\frac{P(B)P(A\mid B)}{P(B)P(A\mid B)+P(\bar B)P(A\mid \bar B)}=\frac{0.5\cdot1}{0.5\cdot1+0.5\cdot\tfrac14}=0.8.$$

(b) 先验 $P(B)=0.2$，有
$$P(B\mid A)=\frac{0.2\cdot1}{0.2\cdot1+0.8\cdot\tfrac14}=0.5.$$

### 3-2 题目

Assume the random variable $X$ has $P\{X=1\}=P\{X=2\}=\frac{1}{2}$. Under the condition $X=i$, the random variable $Y$ follows uniform distribution $U(0, i)$ $i=1,2$. What are the distribution function $F_{Y}(y)$ and the density function $f_{Y}(y)$?

```python
import sympy as sp

y = sp.symbols("y", real=True)
# F_Y = 1/2·F_{U(0,1)} + 1/2·F_{U(0,2)}
# 0<=y<1: y/2 + y/4 = 3y/4 ; 1<=y<2: 1/2 + y/4 ; y>=2: 1
# 密度: 0<=y<1: 3/4 ; 1<=y<2: 1/4
assert sp.integrate(sp.Rational(3, 4), (y, 0, 1)) + sp.integrate(sp.Rational(1, 4), (y, 1, 2)) == 1
# 分段点连续性: F(1-)=3/4, F(1+)=1/2+1/4=3/4; F(2-)=1/2+1/2=1
assert sp.Rational(3, 4) == sp.Rational(1, 2) + sp.Rational(1, 4)
assert sp.Rational(1, 2) + sp.Rational(1, 2) == 1
print("3-2 ok: F 与 f 均正确且连续")
```

### 3-2 答案

\frac{3}{4}y,& 0\le y<1,\\
\frac{1}{2}+\frac{1}{4}y,& 1\le y<2,\\
1,& y\ge 2.
\end{cases}
$$

$$
f_Y(y)=\begin{cases}

### 3-3 题目

If a hen lays eggs $X$ which follows Poisson distribution with the parameter $\lambda$. The probability that each egg transforms into a chick is $p$. Determine the probability of each hen has $k$ chicks $(Y=k)$ such that $Y$ follows Poisson distribution with the parameter $\lambda p$

### 3-3 答案

条件于 $X=n$ 时，$Y\mid X=n\sim\mathrm{Bin}(n,p)$，故对任意 $k\in\mathbb N_0$，

$$
\begin{aligned}
P(Y=k)
&= \sum_{n=k}^{\infty} P(X=n)\binom{n}{k}\, p^k (1-p)^{n-k} \\
&= e^{-\lambda}\frac{(\lambda p)^k}{k!}\sum_{n=k}^{\infty} \frac{[\lambda(1-p)]^{\,n-k}}{(n-k)!} \\
&= e^{-\lambda p}\frac{(\lambda p)^k}{k!}.
\end{aligned}
$$

故 $Y\sim\mathrm{Poisson}(\lambda p)$（随机稀疏化定理）。

### 3-4 题目

Assume $Y=X^{2}$, and the density function of r.v. $X$ is:

$$
f_{X}(x) =
\begin{cases}
c x, & 0<x<2 \\
0, & \text{otherwise}
\end{cases}
$$

(A) What is the constant $c$?

(B) What is the density function $f_{Y}(y)$?

```python
import sympy as sp

x = sp.symbols("x", positive=True)
c = sp.symbols("c", positive=True)
cval = sp.solve(sp.Eq(sp.integrate(c * x, (x, 0, 2)), 1), c)[0]
assert cval == sp.Rational(1, 2)
# Y = X^2: f_Y(y) = f_X(√y)/(2√y) = (√y/2)/(2√y) = 1/4, 0<y<4
yv = sp.symbols("y", positive=True)
fy = cval * sp.sqrt(yv) / (2 * sp.sqrt(yv))
assert sp.simplify(fy - sp.Rational(1, 4)) == 0
assert sp.integrate(sp.Rational(1, 4), (yv, 0, 4)) == 1
(cval, fy)
```

### 3-4 答案

（A）归一化 $1=\int_0^2 c x\,dx=2c$，得 $c=\tfrac12$。

（B）$y=x^2$ 在 $(0,2)$ 上单调，故对 $0<y<4$，
$$
f_Y(y)=\frac{f_X(\sqrt y)}{2\sqrt y}=\frac{\tfrac12\sqrt y}{2\sqrt y}=\tfrac14.
$$

$$
f_Y(y)=\begin{cases}\tfrac14,& 0<y<4,\\ 0,& \text{otherwise}\end{cases},\quad
F_Y(y)=\begin{cases}0,& y\le 0,\\ \tfrac14 y,& 0<y<4,\\ 1,& y\ge 4.\end{cases}
$$

即 $Y\sim U(0,4)$。

Assume the distribution functions of r.v. $X$ and r.v. $Y$ are as follows

F_{X}(x) =\begin{cases}

Furthermore, $P(X=1, Y=1) =\frac{1}{3}$

(b) Are $X$ and $Y$ independent?

(c) When $Y=1$, what is the conditional probability $P(X=k \mid Y=1)$?

```python
from fractions import Fraction as F

# 边际: P(X=0)=1/3, P(X=1)=2/3; P(Y=1)=1/2, P(Y=2)=1/2
# P(X=1,Y=1)=1/3. 由边缘反推其余格子:
assert (F(1, 2) - F(1, 3)) == F(1, 6)   # P(X=0,Y=1)
assert (F(1, 3) - F(1, 6)) == F(1, 6)   # P(X=0,Y=2)
assert (F(2, 3) - F(1, 3)) == F(1, 3)   # P(X=1,Y=2)
# 独立性: 各格 = 边际乘积
assert F(1, 6) == F(1, 3) * F(1, 2) and F(1, 3) == F(2, 3) * F(1, 2)
print("3-5 ok: 全部格子满足独立性")
```

联合分布表：

|      | X=0   | X=1   | 合计   |

| Y=2  | 1/6   | 1/3   | 1/2    |

### 3-6 题目

Assume the joint density function of the two-dimensional random variable $(X, Y)$

$$
f(x, y) = \begin{cases}k e^{-(x+y)}, & 0<x<1,0<y<\infty \\ 0, & \text{otherwise}\end{cases}
$$

(a) Compute the constant $k$

(b) Find the marginal density function of $f_{X}(x), f_{Y}(y)$

(c) Find the distribution function of $Z=\max \{X, Y\}$

```python
import sympy as sp

x, y, u = sp.symbols("x y u", positive=True, real=True)
k = sp.symbols("k", positive=True)
# (a) 归一化
kval = sp.solve(sp.Eq(sp.integrate(sp.integrate(k * sp.exp(-(x + y)), (y, 0, sp.oo)), (x, 0, 1)), 1), k)[0]
assert sp.simplify(kval - 1 / (1 - sp.exp(-1))) == 0
# (b) 边际
fX = sp.integrate(kval * sp.exp(-(x + y)), (y, 0, sp.oo))
# (c) 独立 => F_Z = F_X·F_Y; 0<=u<1: (1-e^{-u})^2/(1-e^{-1}); u>=1: 1-e^{-u}
FZ_mid = (1 - sp.exp(-u))**2 / (1 - sp.exp(-1))
assert sp.simplify(sp.limit(FZ_mid, u, 1, "-") - (1 - sp.exp(-1))) == 0  # 连续
assert sp.simplify(FZ_mid.subs(u, 0)) == 0
(kval, sp.simplify(fX))
```

```python
import numpy as np
import matplotlib.pyplot as plt

us = np.linspace(-0.3, 2.0, 600)
FZ = np.piecewise(us, [us < 0, (us >= 0) & (us < 1), us >= 1],
                  [0, lambda t: (1 - np.exp(-t))**2 / (1 - np.exp(-1)), lambda t: 1 - np.exp(-t)])
fig, ax = plt.subplots(figsize=(5.6, 4.0))
ax.plot(us, FZ, label=r"$F_Z(u)$, $Z=\max\{X,Y\}$")
ax.scatter([1], [1 - np.exp(-1)], color="black", zorder=3, label=r"seam at $u=1$ (continuous)")
ax.grid(alpha=0.3)
ax.legend()
ax.set_title(r"$F_Z$: quadratic piece then pure exponential")
plt.show()
```

### 3-6 答案

（a）归一化 $1=\int_0^1\!\int_0^{\infty} k e^{-(x+y)}\,dy\,dx=k(1-e^{-1})$，得 $k=\dfrac{1}{1-e^{-1}}$。

（b）边际密度：
$$
f_X(x) = \begin{cases}
\dfrac{e^{-x}}{1-e^{-1}}, & 0<x<1,\\
0, & \text{otherwise}
\end{cases}
,\quad
f_Y(y) = \begin{cases}
e^{-y}, & y>0,\\
0, & \text{otherwise}
\end{cases}.
$$

（c）因 $f(x,y)=f_X(x)f_Y(y)$，$X,Y$ 独立，故 $F_Z(u)=F_X(u)F_Y(u)$，其中

$$
F_X(u) = \begin{cases}
0,& u<0,\\
\dfrac{1-e^{-u}}{1-e^{-1}},& 0\le u<1,\\
1,& u\ge 1,
\end{cases}
\quad
F_Y(u)= \begin{cases}
0,& u<0,\\
1-e^{-u},& u\ge 0.
\end{cases}
$$

于是
$$
F_Z(u)=
\begin{cases}
0,& u<0,\\
\dfrac{\left(1-e^{-u}\right)^2}{1-e^{-1}},& 0\le u<1,\\
1-e^{-u},& u\ge 1.
\end{cases}
$$
