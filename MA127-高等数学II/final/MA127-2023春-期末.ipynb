### 1-1 题目

下列哪一个向量与曲线 $r(t)=\left(1+\frac{1}{t^{2}}\right) i+\left(1-\frac{3}{t}\right) j+t^{2} k$ 在点 $P=(2,-2,1)$ 处的切线垂直?

(A) $v=\langle 1,2,1\rangle$  (B) $v=\langle 1,2,2\rangle$

(C) $v=\langle-1,-2,2\rangle$  (D) $v=\langle-1,-2,-2\rangle$

```python
import sympy as sp
from IPython.display import display, Math

t = sp.symbols("t", positive=True)
r = sp.Matrix([1 + 1 / t**2, 1 - 3 / t, t**2])
rp1 = sp.simplify(r.diff(t)).subs(t, 1)
display(Math(r"r'(1)=" + sp.latex(rp1)))
for name, v in [("A", (1, 2, 1)), ("B", (1, 2, 2)), ("C", (-1, -2, 2)), ("D", (-1, -2, -2))]:
    print(name, "r'(1)·v =", rp1.dot(v))
```

### 1-1 答案

**C**

$$
\begin{aligned}
&1+\frac{1}{t^{2}}=2,\quad 1-\frac{3}{t}=-2 \Rightarrow t=1\\
&r^{\prime}(t)=\left\langle-\frac{2}{t^{3}}, \frac{3}{t^{2}}, 2 t\right\rangle,\quad r^{\prime}(1)=\langle-2,3,2\rangle
\end{aligned}
$$

与 $r'(1)$ 垂直的只有 (C)。

### 1-2 题目

函数 $f(x, y, z)=x^{2} y+z^{2}$ 在点 $(1,2,0)$ 处沿向量 $v=\langle 1,2,2\rangle$ 的方向的方向导数是

(A) 12  (B) 6  (C) 4  (D) 2

```python
import sympy as sp
from IPython.display import display, Math

x, y, z = sp.symbols("x y z", real=True)
f = x**2 * y + z**2
v = sp.Matrix([1, 2, 2])
grad = sp.Matrix([sp.diff(f, s) for s in (x, y, z)]).subs({x: 1, y: 2, z: 0})
du = sp.simplify(grad.dot(v.normalized()))
display(Math(r"\nabla f(1,2,0)=" + sp.latex(grad)
             + r",\quad \frac{\vec v}{|\vec v|}=\frac{1}{3}\langle 1,2,2\rangle,\quad D_{\vec v}f=" + sp.latex(du)))
```

### 1-2 答案

**D**

$$
\begin{aligned}
&\nabla f=\left\langle 2xy, x^{2}, 2z\right\rangle,\quad \nabla f(1,2,0)=\langle 4,1,0\rangle\\
&\frac{\vec{v}}{|\vec{v}|}=\frac{1}{3}\langle 1,2,2\rangle,\quad \frac{4+2}{3}=2
\end{aligned}
$$

### 1-3 题目

级数 $\sum\limits_{n=1}^{\infty}\left(1-\dfrac{a}{n}\right)^{n^{2}}$

(A) 对任意实数 $a$ 都收敛  (B) 当 $a \in(-1,1)$ 时收敛

(C) 当 $a$ 为任意正实数时收敛  (D) 当 $a$ 为任意负实数时收敛

```python
import sympy as sp
from IPython.display import display, Math

a, n = sp.symbols("a n", positive=True)
# 根值判别: u_n^{1/n} = (1 - a/n)^n
rho = sp.limit((1 - a / n) ** n, n, sp.oo)
display(Math(r"\sqrt[n]{u_n}=\left(1-\frac{a}{n}\right)^{n}\to " + sp.latex(rho) + r"\ (a>0)"))
```

### 1-3 答案

**C**

当 $a=0$ 时 $\sum\limits_{n=1}^{\infty} 1$ 发散；$a<0$ 时 $1<\left(1-\dfrac{a}{n}\right)^{n^{2}}$，通项不趋于 0，发散。

当 $a>0$ 时，$\lim\limits_{n \rightarrow \infty}\left(1-\dfrac{a}{n}\right)^{n}=\lim\limits_{n \rightarrow \infty}\left(1-\dfrac{a}{n}\right)^{\frac{n}{a} \cdot a}=\dfrac{1}{e^{a}}<1$，由根值判别法收敛。

### 2-1 题目

曲面 $x y z+e^{z+x}=0$ 在点 $(1,1,-1)$ 处的切平面的方程是 $\underline{\quad\quad\quad}$

```python
import sympy as sp
from IPython.display import display, Math

x, y, z = sp.symbols("x y z", real=True)
F = x * y * z + sp.exp(x + z)
gradF = sp.Matrix([sp.diff(F, s) for s in (x, y, z)]).subs({x: 1, y: 1, z: -1})
display(Math(r"\nabla F(1,1,-1)=" + sp.latex(gradF)))
display(Math(r"0\cdot(x-1)-1\cdot(y-1)+2\cdot(z+1)=0\ \Rightarrow\ y-2z-3=0"))
```

### 2-1 答案

$y-2 z-3=0$（即 $y-2z=3$）

$$
\begin{aligned}
&F(x, y, z)=x y z+e^{x+z}\\
&\nabla F=\left\langle yz+e^{x+z}, xz, xy+e^{x+z}\right\rangle\\
&n\big|_{(1,1,-1)}=\langle-1+1,-1,2\rangle=\langle 0,-1,2\rangle\\
&-y+2z=-3\ \text{ or }\ y-2z-3=0
\end{aligned}
$$

```python
import sympy as sp
from IPython.display import display, Math

m = sp.symbols("m", integer=True, positive=True)
m2 = sp.solve(sp.Eq(2 * m + 2, 6), m)[0]
coef = 1 / (m2 * (m2 + 4) * 2**m2)   # x^6 的系数
display(Math(r"n=%d:\ \frac{f^{(6)}(0)}{6!}=[x^6]=\frac{1}{2\cdot 6\cdot 4}" % m2))
display(Math(r"f^{(6)}(0)=6!\cdot\frac{1}{48}=" + sp.latex(sp.factorial(6) * coef)))
```

### 2-2 答案

$$

\end{aligned}
$$

### 2-4 题目

若函数 $y=y(x)$ 的参数方程为 $x=2 t+\ln ^{2} t,\ y=(t+\ln t)^{2},\ t>0$, 则 $\left.\dfrac{d^{2} y}{d x^{2}}\right|_{t=1}=$ $\underline{\quad\quad\quad}$

```python
import sympy as sp
from IPython.display import display, Math

t = sp.symbols("t", positive=True)
X, Y = 2 * t + sp.log(t)**2, (t + sp.log(t))**2
dydx = sp.simplify(Y.diff(t) / X.diff(t))
d2 = sp.simplify(sp.diff(dydx, t) / X.diff(t))
display(Math(r"\frac{dy}{dx}=" + sp.latex(dydx)))
display(Math(r"\frac{d^{2}y}{dx^{2}}\Big|_{t=1}=" + sp.latex(d2.subs(t, 1))))
```

### 2-4 答案

$\dfrac{1}{2}$

$$
\text{Note: } \frac{d y}{d x}=\frac{d y}{d t} \cdot \frac{1}{\frac{d x}{d t}},\quad
\frac{d^{2} y}{d x^{2}}=\frac{d\left(\frac{d y}{d x}\right)}{d t} \cdot \frac{1}{\frac{d x}{d t}}
$$

$$
\frac{d y}{d t}=2(t+\ln t)\left(1+\frac{1}{t}\right),\quad
\frac{d x}{d t}=2+\frac{2\ln t}{t}=\frac{2}{t}(t+\ln t)
$$

首先计算 $\dfrac{d y}{d x}$：$\dfrac{d y}{d x}=\dfrac{2\left(1+\frac{1}{t}\right)}{\frac{2}{t}}=1+t$

然后 $\dfrac{d^{2} y}{d x^{2}}=\dfrac{1}{\frac{2}{t}(t+\ln t)}$，代入 $t=1$ 得 $\dfrac{d^{2} y}{d x^{2}}=\dfrac{1}{2(1+0)}=\dfrac{1}{2}$

### 3 题目

求函数 $f(x, y)=x^{3}+2 y^{2}-3 x-12 y$ 的所有极值

```python
import sympy as sp
from IPython.display import display, Math

x, y = sp.symbols("x y", real=True)
f = x**3 + 2 * y**2 - 3 * x - 12 * y
crit = sp.solve([sp.diff(f, x), sp.diff(f, y)], [x, y], dict=True)
H = sp.hessian(f, (x, y))
for pt in crit:
    d = H.subs({x: pt[x], y: pt[y]}).det()
    fxx = H[0, 0].subs({x: pt[x], y: pt[y]})
    tag = "极小值点" if d > 0 and fxx > 0 else ("极大值点" if d > 0 else "鞍点")
    print(pt, ": D =", d, ", f_xx =", fxx, "->", tag, ", f =", f.subs(pt))
```

### 3 答案

解：

$$
f_{x}(x, y)=3 x^{2}-3=0,\quad f_{y}(x, y)=4 y-12=0
$$

得 $x=\pm 1,\ y=3$，故 $(1,3),(-1,3)$ 为所有可能极值点。

$$
f_{xx}=6x,\quad f_{yy}=4,\quad f_{xy}=0
$$

- for $(1,3)$：$f_{xx}(1,3) f_{yy}(1,3)-f_{xy}^{2}(1,3)=24>0$ 且 $f_{xx}(1,3)>0$，故 $(1,3)$ 为极小值点，$f(1,3)=1+18-3-36=-20$
- for $(-1,3)$：$f_{xx}(-1,3) f_{yy}(-1,3)-f_{xy}^{2}(-1,3)=-24<0$，故 $(-1,3)$ 为鞍点

综上，极小值为 $-20$，在 $(1,3)$ 取到；无极大值。

### 4-1 题目

求级数 $\sum\limits_{n=2}^{\infty} \dfrac{(-1)^{n}(x-1)^{2 n+1}}{(n+2023) \ln n}$ 的收敛域

```python
import sympy as sp
from IPython.display import display, Math

n = sp.symbols("n", integer=True, positive=True)
# 相邻项比值中的系数部分
rho = sp.limit((n + 2023) * sp.log(n) / ((n + 2024) * sp.log(n + 1)), n, sp.oo)
display(Math(r"\rho=\lim\left|\frac{a_{n+1}}{a_n}\right|=(x-1)^{2}\lim\frac{(n+2023)\ln n}{(n+2024)\ln(n+1)}="
             + sp.latex(rho) + r"\cdot(x-1)^{2}"))
```

### 4-1 答案

记 $a_{n}=\dfrac{(-1)^{n}(x-1)^{2 n+1}}{(n+2023) \ln n}$（仅为少写字）

$$
\rho=\lim\limits_{n \rightarrow \infty}\left|\frac{a_{n+1}}{a_{n}}\right|=(x-1)^{2}
$$

当 $0<x<2$ 时，$\rho<1$，$\sum\limits_{n=2}^{\infty} a_{n}$ 收敛；当 $x<0$ 或 $x>2$ 时，$\rho>1$，发散。

现讨论端点：

**for $x=0$：** 记 $b_{n}=\dfrac{(-1)^{n+1}}{(n+2023) \ln n}$，$C_{n}=\dfrac{1}{(n+2023) \ln n}$（同样仅为了少写字）。当 $n \geq 2$ 时 $C_{n}>0$，$\lim\limits_{n \rightarrow \infty} C_{n}=0$ 且 $C_{n}$ 单调减少，由莱布尼茨判别法 $\sum b_{n}$ 收敛，即 $x=0$ 时级数收敛。

**for $x=2$：** 记 $b_{n}^{\prime}=\dfrac{(-1)^{n}}{(n+2023) \ln n}=-b_{n}$，由于 $\sum b_{n}$ 收敛，故 $\sum b_{n}^{\prime}$ 收敛。

综上，收敛域为 $[0,2]$。

### 4-2 题目

$x$ 取哪些值时上述级数绝对收敛, 取哪些值时条件收敛

```python
import sympy as sp
from IPython.display import display, Math

w = sp.symbols("w", positive=True)
# 积分判别法: 1/(2x ln x) 在 [2023, +inf) 上积分发散
inte = sp.integrate(1 / (2 * w * sp.log(w)), (w, 2023, sp.oo))
display(Math(r"\int_{2023}^{+\infty}\frac{dx}{2x\ln x}=" + sp.latex(inte) + r"\ \Rightarrow\ \sum T_n \text{ diverges}"))
```

### 4-2 答案

$(0,2)$ 上绝对收敛，$(-\infty, 0) \cup (2, +\infty)$ 上均发散，因此只需讨论 $0$ 和 $2$ 两个点。

**for $x=0$：** 记 $d_{n}=|b_{n}|=\dfrac{1}{(n+2023) \ln n}$。当 $n>2023$ 时，$0<\dfrac{1}{2n\ln n}<d_{n}$；记 $T_{n}=\dfrac{1}{2n\ln n}$，$f(x)=\dfrac{1}{2x\ln x}$，则 $T_{n}>0$、单调减少且 $f(x)$ 连续，故 $\sum T_{n}$ 与 $\int_{2023}^{+\infty} \dfrac{dx}{2x\ln x}$ 收敛性一致，均发散。故 $\sum|b_{n}|$ 发散，$x=0$ 时级数**条件收敛**。

**for $x=2$：** 此时 $d_{n}^{\prime}=|b_{n}^{\prime}|=|b_{n}|$，同理条件收敛。

综上，在 $(0,2)$ 绝对收敛，在 $x=0$ 和 $x=2$ 条件收敛。

### 5 题目

若上半球 $D: x^{2}+y^{2}+z^{2} \leq 1,\ z \geq 0$ 的密度函数为 $\delta(x, y, z)=z$, 计算此上半球的质心

```python
import sympy as sp
from IPython.display import display, Math

r, phi, theta = sp.symbols("r phi theta", positive=True)
dV = r**2 * sp.sin(phi)
M = sp.integrate(r * sp.cos(phi) * dV, (r, 0, 1), (phi, 0, sp.pi / 2), (theta, 0, 2 * sp.pi))
Mxy = sp.integrate((r * sp.cos(phi))**2 * dV, (r, 0, 1), (phi, 0, sp.pi / 2), (theta, 0, 2 * sp.pi))
zbar = sp.simplify(Mxy / M)
display(Math(r"M=" + sp.latex(M) + r",\quad M_{xy}=" + sp.latex(Mxy) + r",\quad \bar{z}=" + sp.latex(zbar)))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 上半球密度 delta = z：xz 截面上的密度分布 + 质心位置
xg = np.linspace(-1, 1, 240)
zg = np.linspace(0, 1.05, 240)
X, Z = np.meshgrid(xg, zg)
inside = X**2 + Z**2 <= 1
delta = np.where(inside, Z, np.nan)

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
ax = axes[0]
im = ax.pcolormesh(X, Z, delta, cmap="viridis", shading="auto")
fig.colorbar(im, ax=ax, label=r"$\delta = z$")
th = np.linspace(0, np.pi, 200)
ax.plot(np.cos(th), np.sin(th), "k-", lw=1.5)
ax.set_title(r"upper hemisphere, $\delta=z$")
ax.set_xlabel("$x$")
ax.set_ylabel("$z$")
ax.set_aspect("equal")

ax = axes[1]
zs = np.linspace(0, 1, 200)
# 水平薄片质量密度 m(z) = pi*z*(1-z^2), 其加权平均即质心 zbar = 8/15
mz = np.pi * zs * (1 - zs**2)
ax.plot(zs, mz, "b-", lw=1.5, label=r"$m(z)=\pi z(1-z^{2})$")
ax.fill_between(zs, 0, mz, alpha=0.2, color="tab:blue")
ax.axvline(8 / 15, color="r", ls="--", label=r"$\bar{z}=8/15$")
ax.set_title(r"mass per $z$-slice and centroid")
ax.set_xlabel("$z$")
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()
```

### 5 答案

由于积分区域 $D$ 关于 $x=0$, $y=0$ 对称，且 $\iiint_D x\delta\,dv=\iiint_D xz\,dv$ 是 $x$ 的奇函数、$\iiint_D y\delta\,dv=\iiint_D yz\,dv$ 是 $y$ 的奇函数，故 $\iiint_{D} x\delta\,dv=\iiint_{D} y\delta\,dv=0$，即 $\bar{x}=\bar{y}=0$。

现计算 $\bar{z}$，利用球坐标：

$$
\left\{
\begin{array}{llll}
x=r \sin \varphi \cos \theta & \theta: 0 \rightarrow 2 \pi \\
y=r \sin \varphi \sin \theta & \varphi: 0 \rightarrow \frac{\pi}{2} \\
z=r \cos \varphi & r: 0 \rightarrow 1
\end{array}
\right.
$$

$$
\begin{aligned}
M_{xy} &=\iiint_{D} z \delta\, d v\\
&=\int_{0}^{2 \pi} d \theta \int_{0}^{\frac{\pi}{2}} \sin \varphi\, d \varphi \int_{0}^{1} r^{2} \cdot r \cos \varphi \cdot r \cos \varphi\, d r\\
&=2 \pi \int_{0}^{\frac{\pi}{2}} \cos ^{2} \varphi \sin \varphi\, d \varphi \int_{0}^{1} r^{4} d r
=2 \pi \times\left.\left(-\frac{1}{3}\right) \cos ^{3} \varphi\right|_{0} ^{\frac{\pi}{2}} \times \frac{1}{5}
=\frac{2}{15} \pi
\end{aligned}
$$

> 先算好算的，分开算：一起算的话，算对得全分，算错得 0 分；算出一个得一个的分。

$$
\begin{aligned}
M&=\iiint_{D} \delta\, d v\\
&=\int_{0}^{2 \pi} d \theta \int_{0}^{\frac{\pi}{2}} \sin \varphi\, d \varphi \int_{0}^{1} r^{2} \cdot r \cos \varphi\, d r\\
&=2 \pi \times \frac{1}{4} \int_{0}^{\frac{\pi}{2}} \sin \varphi \cos \varphi\, d \varphi
=\frac{\pi}{2} \int_{0}^{\frac{\pi}{2}} \sin \varphi\, d(\sin \varphi)
=\frac{\pi}{4}
\end{aligned}
$$

故 $\bar{z}=\dfrac{M_{xy}}{M}=\dfrac{8}{15}$，质心坐标为 $\left(0,0,\dfrac{8}{15}\right)$。

## 六

### 6 题目

已知向量场 $F=(y \sin z+2) i+(x \sin z) j+(x y \cos z) k$ 是保守场, 求F的势函数, 并计算曲线积分 $\int_{C} F \cdot d r$ .这里C是从点 $A\left(1,-2, \frac{\pi}{3}\right)$ 到点 $B\left(2,1, \frac{\pi}{4}\right)$ 的光滑曲线

```python
import sympy as sp
from IPython.display import display, Math

x, y, z = sp.symbols("x y z", real=True)
P, Q, R = y * sp.sin(z) + 2, x * sp.sin(z), x * y * sp.cos(z)
curlF = sp.Matrix([sp.diff(R, y) - sp.diff(Q, z),
                   sp.diff(P, z) - sp.diff(R, x),
                   sp.diff(Q, x) - sp.diff(P, y)])
# curl F = 0 => F 为保守场
display(Math(r"\nabla\times F=" + sp.latex(curlF) + r"=\vec{0}"))
pot = x * y * sp.sin(z) + 2 * x
display(Math(r"f(x,y,z)=" + sp.latex(pot) + "+C"))
val = sp.simplify(pot.subs({x: 2, y: 1, z: sp.pi / 4}) - pot.subs({x: 1, y: -2, z: sp.pi / 3}))
display(Math(r"\int_C F\cdot dr=f(B)-f(A)=" + sp.latex(val)))
```

### 6 答案

解：设 $f(x, y, z)$ 为 F 的势函数，则有

$$
\left\{
\begin{array}{l}
(1): f_x(x, y, z)=y \sin z+2\\
(2): f_y(x, y, z)=x \sin z\\
(3): f_z(x, y, z)=x y \cos z
\end{array}
\right.
$$

由 $(1)$ 积分得 $(4): f(x, y, z)=x y \sin z+2 x+g(y, z)$

$(4)$ 代入 $(2)$：$(5): x \sin z+g_y(y, z)=x \sin z$，so $g(y,z)=g(z)$

$(5)$ 代入 $(4)$：$(6): f(x, y, z)=x y \sin z+2 x+g(z)$

$(6)$ 代入 $(3)$：$x y \cos z+g^{\prime}(z)=x y \cos z$，so $g(z)=C$

故 F 的势函数 $f(x, y, z)=x y \sin z+2 x+C$

$$
\int_c \vec{F} \cdot d \vec{r}=f\left(2,1, \frac{\pi}{4}\right)-f\left(1,-2, \frac{\pi}{3}\right)
=2 \times \frac{\sqrt{2}}{2}+4-\left(-2 \times \frac{\sqrt{3}}{2}+2\right)=\sqrt{2}+\sqrt{3}+2
$$

## 七

### 7 题目

向量场 $F(x, y, z)=\left(z+x y^{2}\right) i+\left(2 y+y^{3}\right) j-4 z y^{2} k$ 的定义域 $V$ 是夹在如下曲面 $S_{1}$ 和 $S_{2}$ 之间的闭区域, 这里

$S_{1}:=\left\{(x, y, z): z=\left(x^{2}+y^{2}\right)^{2}-1\right\}, \quad S_{2}:=\left\{(x, y, z): z=4-4\left(x^{2}+y^{2}\right)\right\}$

使用散度定理来计算向量场通过 V 的边界从内向外的通量(flux)

```python
import sympy as sp
from IPython.display import display, Math

x, y, z = sp.symbols("x y z", real=True)
div = sp.diff(z + x * y**2, x) + sp.diff(2 * y + y**3, y) + sp.diff(-4 * z * y**2, z)
display(Math(r"\operatorname{div}F=" + sp.latex(sp.simplify(div))))

r, theta = sp.symbols("r theta", positive=True)
zc = r**4 - 1
ztop = 4 - 4 * r**2
flux = sp.integrate(sp.integrate(r * sp.integrate(div.subs({x: r * sp.cos(theta), y: r * sp.sin(theta)}),
                                                 (z, zc, ztop)), (r, 0, 1)), (theta, 0, 2 * sp.pi))
display(Math(r"\text{Flux}=\iiint_R \operatorname{div}F\,dv=" + sp.latex(flux)))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# S1: z = (x^2+y^2)^2 - 1 (lower bowl), S2: z = 4 - 4(x^2+y^2) (upper paraboloid)
r = np.linspace(0, 1, 60)
R, T = np.meshgrid(r, np.linspace(0, 2 * np.pi, 80))
X, Y = R * np.cos(T), R * np.sin(T)
Z1, Z2 = (X**2 + Y**2) ** 2 - 1, 4 - 4 * (X**2 + Y**2)

fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(projection="3d")
ax.plot_surface(X, Y, Z1, alpha=0.5, color="tab:blue")
ax.plot_surface(X, Y, Z2, alpha=0.5, color="tab:orange")
# 交线: x^2+y^2=1, z=0
tline = np.linspace(0, 2 * np.pi, 100)
ax.plot(np.cos(tline), np.sin(tline), 0, "k-", lw=2)
ax.set_xlabel("$x$")
ax.set_ylabel("$y$")
ax.set_zlabel("$z$")
ax.set_title(r"$V$ between $S_1$ (blue) and $S_2$ (orange), $r=1, z=0$")
plt.show()
```

### 7 答案

不好画也可以不画，只需搞清楚积分区域 $S_1, S_2$ 谁在上、谁在下。

**7-1 解交线、找投影**

$$
z=\left(x^{2}+y^{2}\right)^{2}-1=4-4\left(x^{2}+y^{2}\right) \Rightarrow\left\{\begin{array}{l}x^{2}+y^{2}=1 \\ z=0\end{array}\right.
$$

因此积分区域在 $xy$ 上的投影为 $D: x^{2}+y^{2} \leq 1$。

**7-2 定 $S_1, S_2$ 谁上谁下**

当 $x^{2}+y^{2} \leq 1$ 时 $S_{1} \leq S_{2}$（$S_2$ 在上），由散度定理：

$$
\text{Flux}=\unicode{8751}{\partial R} \vec{F} \cdot \vec{n}\, d \sigma=\iiint_R \operatorname{div} \vec{F}\, d v
=\iint_D d A \int_{\left(x^{2}+y^{2}\right)^{2}-1}^{4-4\left(x^{2}+y^{2}\right)} \operatorname{div} \vec{F}\, d z
$$

$$
\operatorname{div} \vec{F}=y^{2}+2+3 y^{2}-4 y^{2}=2
$$

利用极坐标计算：

$$
\left\{\begin{array}{ll}
x=r \cos \theta & \theta: 0 \rightarrow 2 \pi \\
y=r \sin \theta & r: 0 \rightarrow 1 \\
z=z & z: r^{4}-1 \rightarrow 4-4 r^{2}
\end{array}\right.
$$

$$
\text{Flux}=2 \int_{0}^{2 \pi} d \theta \int_{0}^{1} r\, d r \int_{r^{4}-1}^{4-4 r^{2}} d z
=4 \pi \int_{0}^{1} r\left(5-4 r^{2}-r^{4}\right) d r
=4 \pi\left(\frac{5}{2}-1-\frac{1}{6}\right)=2 \pi \times \frac{8}{3}=\frac{16}{3} \pi
$$

## 八

### 8 题目

设曲面 $S$ 是椭圆抛物面 $z=x^{2}+4 y^{2}$ 在平面 $z=1$ 下方的部分, 取 $S$ 的内法向量 $n$(即正 $k$ 分量的法向量)为定向, 计算 $\iint_{S} \nabla \times F \cdot n\, d \sigma$,

这里 $F=\left(y+x^{2} \ln \left(x^{4}+1\right)\right) i+\left(e^{\sin y}-x z\right) j+\left(x z^{2}+\cos \left(z^{2}+1\right)\right) k$

> 原卷此处附图: 椭圆抛物面 $z=x^{2}+4 y^{2}$ 位于平面 $z=1$ 下方, 边界曲线为 $C: x^{2}+4 y^{2}=1$, 图中标出了内法向量 $n$ 的方向.

```python
import sympy as sp
from IPython.display import display, Math

x, y, z = sp.symbols("x y z", real=True)
P = y + x**2 * sp.log(x**4 + 1)
Q = sp.exp(sp.sin(y)) - x * z
R = x * z**2 + sp.cos(z**2 + 1)
curlF = sp.Matrix([sp.diff(R, y) - sp.diff(Q, z),
                   sp.diff(P, z) - sp.diff(R, x),
                   sp.diff(Q, x) - sp.diff(P, y)])
display(Math(r"\nabla\times F=" + sp.latex(curlF)))
# Sigma: z=1 平面上椭圆 x^2 + 4y^2 <= 1 内部, n=(0,0,1); 被积函数 = (curl·n)|_{z=1} = -z-1 = -2
s_ellipse = sp.pi * 1 * sp.Rational(1, 2)
display(Math(r"\iint_{\Sigma}(\nabla\times F)\cdot n\,d\sigma=-2\cdot S_{\Sigma}=-2\cdot\pi\cdot 1\cdot\frac{1}{2}="
             + sp.latex(-2 * s_ellipse)))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 抛物面 z = x^2 + 4y^2 被 z=1 截断; 边界 C: x^2 + 4y^2 = 1, z=1
X, Y = np.meshgrid(np.linspace(-1, 1, 80), np.linspace(-0.6, 0.6, 80))
Z = X**2 + 4 * Y**2
Zp = np.where(Z <= 1, Z, np.nan)

fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(projection="3d")
ax.plot_surface(X, Y, Zp, alpha=0.55, cmap="viridis")
t = np.linspace(0, 2 * np.pi, 200)
ax.plot(np.cos(t), 0.5 * np.sin(t), 1, "r-", lw=2.5, label=r"$C: x^2+4y^2=1,\ z=1$")
ax.set_xlabel("$x$")
ax.set_ylabel("$y$")
ax.set_zlabel("$z$")
ax.set_title(r"elliptic paraboloid $z=x^2+4y^2$ and curve $C$")
ax.legend()
plt.show()
```

### 8 答案

解：利用斯托克斯公式

$$
\iint_{S} \nabla \times \vec{F} \cdot \vec{n}\, d \sigma=\oint_{C} \vec{F} \cdot d \vec{r}
$$

其中 $C$ 为 $\left\{\begin{array}{c}x^{2}+4 y^{2}=z \\ z=1\end{array}\right.$ 的交线，从上往下看为逆时针。再化为平面积分：

$$
\oint_{C} \vec{F} \cdot d \vec{r}=\iint_{\Sigma} \nabla \times \vec{F} \cdot \vec{n}\, d \sigma
$$

其中 $\Sigma$ 为 $z=1$ 在 $x^{2}+4 y^{2}=1$ 内部的区域，方向为 $(0,0,1)$（注意与 $S$ 的定向匹配）。

$$
\begin{aligned}
&\nabla \times \vec{F}=\left|\begin{array}{ccc}
i & j & k \\
\frac{\partial}{\partial x} & \frac{\partial}{\partial y} & \frac{\partial}{\partial z} \\
y+x^{2} \ln \left(x^{4}+1\right) & e^{\sin y}-x z & x z^{2}+\cos \left(z^{2}+1\right)
\end{array}\right|=\langle x,-z^{2},-z-1\rangle\\
&\vec{n}=\langle 0,0,1\rangle\\
&\text{original}=\iint_{\Sigma}(-z-1)\, d \sigma,\quad \text{since } \Sigma \text{ is on } z=1 \text{ so } -z-1=-2
\end{aligned}
$$

$$
\frac{x^{2}}{1^{2}}+\frac{y^{2}}{\left(\frac{1}{2}\right)^{2}}=1 \quad \Rightarrow \quad S_{\Sigma}=\pi \times 1 \times \frac{1}{2}
$$

$$
\iint_{S} \nabla \times \vec{F} \cdot \vec{n}\, d \sigma=-2 \times \frac{\pi}{2}=-\pi
$$
