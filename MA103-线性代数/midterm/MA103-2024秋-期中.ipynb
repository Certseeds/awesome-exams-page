### 1-1 题目

Suppose $A$ is an $m \times n$ real matrix with $m<n$, and the equation $A x=b$ has a solution for any $m$-dimensional real column vector $b$. Which of the following assertions must be true?

(A) $A x=b$ has a unique solution for every $b \in \mathbb{R}^{m}$

(B) $A^{\top} x=d$ has a solution for any $d \in \mathbb{R}^{n}$

(C) $N(A)=\{0\}$

(D) A has a right inverse

```
import sympy as sp
from IPython.display import display, Math

# m < n, 且 A x = b 对任意 b 有解 => rank(A) = m（行满秩，列多于行）
A = sp.Matrix([[1, 0, 0, 0],
               [0, 1, 0, 0]])   # 2 x 4 行满秩的例子
b = sp.Matrix([3, -1])
xx = sp.symbols('x0:4')
particular = sp.linsolve((A, b), xx)
display(Math(r'Ax=b \text{ 的解集（含自由变量，解不唯一）}: ' + sp.latex(particular)))
display(Math(r'\operatorname{rank}(A) = %d < n = %d \Rightarrow N(A) \neq \{0\}'
             % (A.rank(), A.shape[1])))
```

### 1-1 答案

Answer: D

解析：$A \in \mathbb{R}^{m \times n},\ \forall b \in \mathbb{R}^m,\ Ax=b$ 有解 $\Rightarrow \operatorname{rank}(A)=m$（即 $C(A)=\mathbb{R}^m$，$R(A,b)=R(A)$）。

- A：行满秩但 $n>m$，$N(A)$ 非零，解不唯一。如 $A=\left[L \mid \alpha_1 \cdots \alpha_m\right]$。
- B：需 $\forall d \in \mathbb{R}^n$，$A^{\top}x=d$ 有解 $\Rightarrow \operatorname{rank}(A)=n$，而这里 $m<n$，不成立。
- C：$Ax=\sum_{i=1}^{n} \alpha_i x_i$，$n>m$ 必有自由列，$N(A)\neq\{0\}$。
- D：$A_{m\times n}$ 有右逆 $\iff \operatorname{rank}(A)=m$。取 $A$ 的 $m$ 个线性无关列拼成可逆的 $A'_{m\times m}$，则 $A$ 的右逆为 $\binom{A'^{-1}}{0}$ 形式的 $n\times m$ 矩阵 $B$，满足 $AB=I_m$。

### 1-2 题目

Suppose we have matrices $A, B$ satisfying $E A=B$ for some invertible matrix $E$. Which of the following assertions must be true?

(A) $N(A)=N(B)$

(B) $C(A)=C(B)$

(C) $N(A^{\top})=N(B^{\top})$

(D) $A^{\top} A=B^{\top} B$

```
import sympy as sp
from IPython.display import display, Math

# EA = B（E 可逆，左乘行变换）：A、B 行等价 => N(A) = N(B)
# 构造反例说明 B、C、D 不必成立
A = sp.Matrix([[1, 0, 0],
               [0, 1, 0],
               [0, 0, 0]])
E = sp.Matrix([[1, 0, 0],
               [0, 1, 0],
               [-1, 0, 1]])
B = E * A
display(Math(r'E = ' + sp.latex(E) + r',\quad B = EA = ' + sp.latex(B)))
display(Math(r'N(A)=N(B):\; ' + str(A.nullspace() == B.nullspace())))
display(Math(r'C(A)=C(B):\; ' + str(A.columnspace() == B.columnspace())
             + r',\quad N(A^{\top})=N(B^{\top}):\; '
             + str(A.T.nullspace() == B.T.nullspace())))
display(Math(r'A^{\top}A = ' + sp.latex(A.T*A) + r' \neq B^{\top}B = ' + sp.latex(B.T*B)))
```

### 1-2 答案

Answer: A

解析：

- A：$Ax=0 \Rightarrow EAx=0 \Rightarrow Bx=0$，又 $E$ 可逆可反推，故 $N(A)=N(B)$。
- B：反例 let $A=\begin{bmatrix}1&0&0\\0&1&0\\0&0&0\end{bmatrix}$，$B=\begin{bmatrix}1&0&0\\0&1&0\\-1&0&0\end{bmatrix}$（也许 `B[2][1]` 是 $-1$，原答案此处模糊），列空间不同。
- C：$A^{\top}x=0 \not\Leftrightarrow B^{\top}x=0$，因 $B^{\top}=A^{\top}E^{\top}$，零空间一般不同。
- D：$B^{\top}B=A^{\top}E^{\top}EA \neq A^{\top}A$（除非 $E$ 正交）。

### 1-3 题目

For any $m \times n$ matrix $A$ with reduced row echelon form $U$. Which of the following assertions must be true?

(A) $C(A)=C(A^{\top} A)$

(B) $\operatorname{rank}(A)=\operatorname{rank}(A^{\top} A)$

(C) $C(A)=C(U)$

(D) If $C(A)=\mathbb{R}^{m}$, then $A^{\top} A$ is invertible

```
import sympy as sp
from IPython.display import display, Math

# rank(A) = rank(A^T A)：Ax=0 与 A^T A x = 0 同解
A = sp.Matrix([[1, 2],
               [2, 4],
               [1, 1]])   # m=3 > n=2 的例子
display(Math(r'A = ' + sp.latex(A)))
display(Math(r'\operatorname{rank}(A) = %d, \quad \operatorname{rank}(A^{\top}A) = %d'
             % (A.rank(), (A.T*A).rank())))
# D 的反例：m < n 时 A^T A 是 n x n 的，秩最多为 m < n，必不可逆
A2 = sp.Matrix([[1, 0, 0], [0, 1, 0]])   # C(A2) = R^2, m=2 < n=3
display(Math(r'C(A)=\mathbb{R}^{m} \text{ 但 } A^{\top}A = ' + sp.latex(A2.T*A2)
             + r' \text{ 不可逆（} n>m \text{）}'))
```

### 1-3 答案

Answer: B

解析：

- A：$C(A) \subseteq C(A^{\top}A)$ 方向一般不成立（$A^{\top}A$ 的列是 $A$ 的列的线性组合，故其实 $C(A^{\top}A)\subseteq C(A)$；但行变换会改变列空间，反例见对称性）——结论不必真。
- B：$Ax=0 \Rightarrow A^{\top}Ax=0$；反之 $A^{\top}Ax=0 \Rightarrow x^{\top}A^{\top}Ax=0 \Rightarrow \|Ax\|^2=0 \Rightarrow Ax=0$，两方程组同解，故秩相等。
- C：行变换（化 $U$）保持行空间 $C(A^{\top})$，不保持列空间 $C(A)$。
- D：当 $n>m$ 时 $A^{\top}A$ 是 $n\times n$ 矩阵，秩 $\le m < n$，必不可逆。

### 1-5 题目

Let $A$ be an $n \times n$ matrix $(n>1)$ such that $A=A^{2}$, and $I$ be the $n \times n$ identity matrix, then

(A) $\operatorname{rank}(A)+\operatorname{rank}(A-I)>n$

(B) $\operatorname{rank}(A)+\operatorname{rank}(A-I)<n$

(C) $\operatorname{rank}(A)+\operatorname{rank}(A-I)=n$

(D) $\operatorname{rank}(A)+\operatorname{rank}(A-I)=n-1$

```
import sympy as sp
from IPython.display import display, Math

# 幂等矩阵 A^2 = A：特征值只能是 0 或 1，秩 = 特征值 1 的重数
for diag_vals in ([1, 1, 0], [1, 0, 0], [0, 0, 0], [1, 1, 1]):
    P = sp.Matrix(sp.randMatrix(3, 3, min=-3, max=3, seed=sum(diag_vals) + 5))
    A = P * sp.diag(*diag_vals) * P.inv()   # A = P diag P^{-1}，幂等且可对角化
    r = A.rank() + (A - sp.eye(3)).rank()
    display(Math(r'A \text{ 的特征值为 } %s: \; \operatorname{rank}(A)+\operatorname{rank}(A-I) = %d'
                 % (str(diag_vals), r)))
```

### 2-2 题目

Let $A=\left[\begin{array}{ccc}1 & 2 & 3 \\ 2 & -1 & 1 \\ 3 & 1 & 3\end{array}\right]$. The LDU factorization of $A$ has $L=$ ______

```
import sympy as sp
from IPython.display import display, Math

A = sp.Matrix([[1, 2, 3], [2, -1, 1], [3, 1, 3]])
# 手工 LDU：消元记录乘数得 L，对角元提出得 D，行归一化得 U
U = A.copy(); L = sp.eye(3)
for k in range(3):
    for i in range(k+1, 3):
        f = sp.Rational(U[i, k], U[k, k])
        L[i, k] = f
        U[i, :] = U[i, :] - f*U[k, :]
D = sp.diag(*[U[i, i] for i in range(3)])
U = sp.Matrix([[U[i, j]/U[i, i] for j in range(3)] for i in range(3)])
display(Math(r'L = ' + sp.latex(L) + r',\quad D = ' + sp.latex(D) + r',\quad U = ' + sp.latex(U)))
display(Math(r'LDU = A:\; ' + str(L*D*U == A)))
```

### 2-2 答案

$$
L=\left[\begin{array}{ccc}
1 & 0 & 0 \\
2 & 1 & 0 \\
3 & 1 & 1
\end{array}\right]
$$

### 2-4 题目

Let $A$ be a $2024 \times 2025$ real matrix with $\operatorname{dim} N(A)=11$, then $\operatorname{dim} N(A^{\top})=$ ______

```
import sympy as sp
from IPython.display import display, Math

m, n, dim_NA = 2024, 2025, 11
r = n - dim_NA                       # rank(A) = n - dim N(A)
left_null = m - r                    # dim N(A^T) = m - rank(A)
display(Math(r'\operatorname{rank}(A) = n - \dim N(A) = %d - %d = %d'
             % (n, dim_NA, r)))
display(Math(r'\dim N(A^{\top}) = m - \operatorname{rank}(A) = %d - %d = %d'
             % (m, r, left_null)))
```

### 2-4 答案

$10$

对 $A \in \mathbb{R}^{m \times n}$：

- $\dim N(A)=n-\operatorname{rank}(A)$
- $\dim C(A)=\operatorname{rank}(A)$，$\dim C(A^{\top})=\operatorname{rank}(A)$
- $\dim N(A^{\top})=m-\operatorname{rank}(A)$

本题 $m=2024,\ n=2025$，$\operatorname{rank}(A)=2025-11=2014$，故 $\dim N(A^{\top})=2024-2014=10$。

```
import sympy as sp
from IPython.display import display, Math

A = sp.Matrix([[0, 1, 0], [1, 0, -1], [0, 1, 0]])
X = (sp.eye(3) - A**2).inv() * (sp.eye(3) - A).inv()
# X - X A^2 - A X + A X A^2 = (I - A) X (I - A^2) = I
# 注意 X = (I-A^2)^{-1} (I-A)^{-1} 中因式顺序：展开 (I-A)X(I-A^2)=I 得 X = (I-A)^{-1}(I-A^2)^{-1}
X2 = (sp.eye(3) - A).inv() * (sp.eye(3) - A**2).inv()
display(Math(r'(I-A)X(I-A^{2}) = ' + sp.latex((sp.eye(3)-A)*X2*(sp.eye(3)-A**2))))
display(Math(r'X = (I-A)^{-1}(I-A^{2})^{-1} = [(I-A^{2})(I-A)]^{-1} = ' + sp.latex(X2)))
```

Consider the following subspace of $\mathbb{R}^{3}$:

V=\left\{
\left.
\left[
\begin{array}{l}
x_{1} \\

### 5-a 题目

Show that:

$$
V_{1}=
\left[
\begin{array}{c}
-1 \\
1 \\
0
\end{array}
\right],
V_{2}=
\left[
\begin{array}{c}
-1 \\
0 \\
1
\end{array}
\right]
$$

is a basis of $V$

```
import sympy as sp
from IPython.display import display, Math

# V: x1 + x2 + x3 = 0 的零空间；验证 v1, v2 属于 V 且线性无关
c1, c2 = sp.symbols('c1 c2')
v1 = sp.Matrix([-1, 1, 0]); v2 = sp.Matrix([-1, 0, 1])
display(Math(r'v_1+v_1 的分量和 = ' + str(sum(v1)) + r',\quad v_2 \text{ 的分量和} = ' + str(sum(v2))))
M = sp.Matrix.hstack(v1, v2)
display(Math(r'[v_1\;v_2] \text{ 秩} = %d \Rightarrow \text{线性无关，而 } \dim V = 2'
             % M.rank()))
display(Math(r'V = N([1\;1\;1]) = \mathrm{span}\{v_1, v_2\}'))
```

### 5-a 答案

（1）$v_{1}, v_{2}$ 线性无关；（2）$v_{1}, v_{2} \in V$（分量和均为 0）。

而 $\dim V = 3 - 1 = 2$，故 $\operatorname{span}(v_{1}, v_{2}) = V$，即 $v_1, v_2$ 是 $V$ 的基。

（解 $x_{1}+x_{2}+x_{3}=0$ 也可得 $V = \mathrm{span}\{(-1,1,0)^{\top},(-1,0,1)^{\top}\}$。）

## 七

Prove the following two independent statements

### 7-a 题目

Let $A$ be an $m \times n$ real matrix. Suppose $v_{1}, v_{2}, \cdots, v_{s}$ is a basis of $C(A^{\top})$ and $w_{1}, w_{2}, \cdots, w_{t}$ is a basis of $N(A)$. Show that: $s+t=n$ and

$$
v_{1}, v_{2}, \ldots, v_{s}, w_{1}, w_{2}, \cdots, w_{t}
$$

is a basis of $\mathbb{R}^{n}$

```
import sympy as sp
from IPython.display import display, Math

# 数值例：A 3x4，验证 dim C(A^T) + dim N(A) = n，且两组基拼成的 4x4 矩阵可逆
A = sp.Matrix([[1, 0, 2, 0], [1, 1, 3, 0], [1, 0, 2, 1]])
Vs = sp.Matrix.hstack(*A.T.columnspace())   # 4 x s
Ws = sp.Matrix.hstack(*A.nullspace())       # 4 x t
display(Math(r's = %d,\quad t = %d,\quad s + t = %d = n'
             % (Vs.shape[1], Ws.shape[1], Vs.shape[1] + Ws.shape[1])))
VW = sp.Matrix.hstack(Vs, Ws)
display(Math(r'\det\,[\,v_1 \cdots v_s\; w_1 \cdots w_t\,] = ' + sp.latex(VW.det())
             + r' \neq 0 \Rightarrow \text{构成 } \mathbb{R}^{n} \text{ 的基}'))
```

### 7-a 答案

$C(A^{\top})$ 的基：$v_{1}, \ldots, v_{s}$；$N(A)$ 的基：$w_{1}, \ldots, w_{t}$。

**先证 $s+t=n$：**

$$
t=\dim N(A)=n-\operatorname{rank}(A)=n-\dim C(A^{\top})=n-s
$$

**再证拼接后线性无关：**

写 $A=\left[\alpha_{1}\ \alpha_{2} \ldots \alpha_{m}\right]$（按行分块），则 $C(A^{\top})=\mathrm{span}(\alpha_{1}, \cdots, \alpha_{m})$，而

$$
N(A)=\{x: Ax=0\}=\{x: \alpha_{i}^{\top} x=0,\ i=1,2, \ldots, m\}
$$

故对任意 $i, j$ 有 $v_{i}^{\top} w_{j}=0$，即 $N(A) \perp C(A^{\top})$。

设

$$
k_{1} v_{1}+\cdots+k_{s} v_{s}+k_{s+1} w_{1}+\cdots+k_{n} w_{t}=0
$$

记 $v=\sum k_i v_i \in C(A^{\top})$，$w=\sum k_j w_j \in N(A)$，则 $v+w=0$。两边与 $v$ 作内积：

$$
v^{\top}(v+w)=v^{\top}v=0 \Rightarrow v=0
$$

从而 $k_{1}, \ldots, k_{s}=0$；similarly $k_{s+1}, \cdots, k_{n}=0$。故线性无关，共 $s+t=n$ 个，构成 $\mathbb{R}^{n}$ 的基。

### 7-b 题目

Let $A$ be an $m \times n$ real matrix with rank $m$. Show that $A A^{\top}$ is invertible

```
import sympy as sp
from IPython.display import display, Math

# 数值例：行满秩 A，验证 A A^T 可逆
A = sp.Matrix([[1, 0, 2], [0, 1, 3]])   # 2 x 3，rank = 2 = m
display(Math(r'\operatorname{rank}(A) = %d = m,\quad \det(AA^{\top}) = '
             % A.rank() + sp.latex((A*A.T).det())))
display(Math(r'AA^{\top} = ' + sp.latex(A*A.T) + r',\quad (AA^{\top})^{-1} = ' + sp.latex((A*A.T).inv())))
```

### 7-b 答案

$A \in \mathbb{R}^{m \times n}$，$\operatorname{rank}(A)=m \Rightarrow \operatorname{rank}(A^{\top})=m$。

$AA^{\top}$ 是 $m \times m$ 矩阵，只需证其秩为 $m$：

$$
\operatorname{rank}(AA^{\top})=m-\dim N(AA^{\top})=m-\dim N(A^{\top})=m-(m-\operatorname{rank}(A))=\operatorname{rank}(A)=m
$$

（其中用到 $N(AA^{\top})=N(A^{\top})$：若 $A^{\top}x=0$ 则显然 $AA^{\top}x=0$；反之 $AA^{\top}x=0 \Rightarrow x^{\top}AA^{\top}x=\|A^{\top}x\|^{2}=0 \Rightarrow A^{\top}x=0$。）

故 $AA^{\top}$ 为 $m$ 阶满秩方阵，即可逆。
